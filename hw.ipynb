7. Հաշվել 1-ից մինչև N թվերի գումարը։


```python
def sum_of_numbers(n):
    result = (n*(n+1))//2
    return result
#with while loop
user_input = int(input("enter a number: "))
total = 0
while user_input>0:
    total+= user_input
    user_input -= 1
print("sum of numbers is: ", total)
```

8. Հաշվել 1-ից մինչև N բոլոր զույգ թվերի գումարը։


```python
def sum_of_even_numbers(n):
    total = 0
    while(n>0):
        if n%2 ==0:
            total+=n
        n -=1
    return  total

print("sum of even numbers is: ", sum_of_even_numbers(10))
```

12. Հաշվել թվի թվանշանների գումարը։
58394
↓
29


```python
def sum_of_digits(n):
    total = 0
    while(n>0):
        total += n%10
        n = n//10
    return total
print(sum_of_digits(58394))
```

13. Հաշվել տողի մեջ միայն ձայնավոր տառերի քանակը։
(Անգլերեն ձայնավորներ՝ a, e, i, o, u)


```python
def count_of_volws(n):
    count = 0
    n = n.lower()
    for i in n:
        if i in "aeiou":
            count +=1
    return count

print(count_of_volws("Hello World"))
```

20. Տրված է tuple։
numbers = (4, 7, 1, 8, 2)
Գտնել ամենափոքր տարրը։


```python
numbers = (4, 7, 1, 8, 2)
smallest = numbers[0]
for i in range(1,len(numbers)):
    if numbers[i]<smallest:
        smallest = numbers[i]
print(smallest)
```
